```python
import torch
from torch.utils.data import Dataset, DataLoader
from torch.utils.data import random_split
from transformers import AutoTokenizer
import os
from PIL import Image, UnidentifiedImageError
import torchvision.transforms as T
import torchvision.transforms.functional as F

class ResizePadToSquare:
    def __init__(self, target_size=224, fill=255):
        self.target_size = target_size
        self.fill = fill

    def __call__(self, image: Image.Image):
        # Convert to grayscale if not L or RGB
        if image.mode not in ['L', 'RGB']:
            image = image.convert('L')

        w, h = image.size
        scale = self.target_size / max(w, h)
        new_w, new_h = int(w * scale), int(h * scale)
        image = F.resize(image, (new_h, new_w))

        # Padding to square
        if image.mode == 'L':
            new_image = Image.new("L", (self.target_size, self.target_size), self.fill)
        else:
            new_image = Image.new("RGB", (self.target_size, self.target_size), (self.fill,) * 3)

        paste_x = (self.target_size - new_w) // 2
        paste_y = (self.target_size - new_h) // 2
        new_image.paste(image, (paste_x, paste_y))

        # Convert grayscale to RGB if needed
        if new_image.mode == 'L':
            new_image = new_image.convert('RGB')

        return new_image


transform = T.Compose([
    ResizePadToSquare(target_size=224),  # your custom transform
    T.ToTensor(),                        # [0, 255] to [0.0, 1.0]
    T.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])  # normalize to [-1, 1]
])

class PatentDataset(Dataset):
    def __init__(self, df, tokenizer, image_dir, transform, max_words=150, batch_size=16, max_length=512, label_column="cit_5yr_window"):
        self.df = df
        self.tokenizer = tokenizer
        self.max_words = max_words
        self.batch_size = batch_size
        self.max_length = max_length
        self.label_column = label_column

        # Group all images by patent ID
        self.transform = transform
        self.patent_to_images = {}
        for filename in os.listdir(image_dir):
            if filename.endswith('.png'):
                patent_id = filename.split('_')[0]
                path = os.path.join(image_dir, filename)
                self.patent_to_images.setdefault(patent_id, []).append(path)
        
        # Patent ID list is the dataset index
        self.patent_ids = list(self.patent_to_images.keys())
        self.valid_indices = self._find_valid_indices()

    def chunk_text(self, text):
        words = text.split()
        return [' '.join(words[i:i + self.max_words]) for i in range(0, len(words), self.max_words)]

    def tokenize_chunks(self, text):
        chunks = self.chunk_text(text)
        # Tokenize all chunks at once
        inputs = self.tokenizer(chunks, padding='max_length', truncation=True, 
                              return_tensors='pt', max_length=self.max_length)
        return {
            "input_ids": inputs["input_ids"],
            "attention_mask": inputs["attention_mask"]
        }
        
    def _find_valid_indices(self):
        valid_indices = []
        for idx in range(len(self.df)):
            row = self.df.iloc[idx]
            patent_id = row["publication_number"]
            if (patent_id in self.patent_to_images and 
                len(self.patent_to_images[patent_id]) > 0):
                valid_indices.append(idx)
        return valid_indices
    
    def __len__(self):
        return len(self.valid_indices)

    def __getitem__(self, idx):
        actual_idx = self.valid_indices[idx]
        row = self.df.iloc[actual_idx]
        patent_id = row["publication_number"]
        image_paths = self.patent_to_images[patent_id]

        images = []
        for path in sorted(image_paths):  # sort for consistency
            try:
                img = Image.open(path).convert("RGB")  # Always force RGB
            except UnidentifiedImageError:
                continue
            if self.transform:
                img = self.transform(img)
            images.append(img)

        # Return stacked images [num_images, 3, H, W] and the patent ID
        image_tensor = torch.stack(images)
        return {
            "title": self.tokenize_chunks(row["title_localized"]),
            "abstract": self.tokenize_chunks(row["abstract_localized"]),
            "claims": self.tokenize_chunks(row["claims_localized"]),
            "description": self.tokenize_chunks(row["description_localized"]),
            "images": image_tensor,
            "label": torch.tensor(row[self.label_column], dtype=torch.float)
        }


# Example Usage
model_name = 'AI-Growth-Lab/PatentSBERTa'


# Example Usage
tokenizer = AutoTokenizer.from_pretrained(model_name)
dataset = PatentDataset(merged_df, tokenizer, 'data/patent_image/2010',transform, batch_size=16)
# Set your split proportions
train_ratio = 0.8
val_ratio = 0.1
test_ratio = 0.1

# Calculate lengths
dataset_size = len(dataset)
train_size = int(train_ratio * dataset_size)
val_size = int(val_ratio * dataset_size)
test_size = dataset_size - train_size - val_size  # ensure total matches

# Split the dataset
train_dataset, val_dataset, test_dataset = random_split(dataset, [train_size, val_size, test_size])
```

```python
import torch
import torch.nn as nn
from transformers import ViTModel

class WeightedImagePooling(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.attn = nn.Linear(input_dim, 1)

    def forward(self, image_embeddings):  # [N, D]
        weights = self.attn(image_embeddings)  # [N, 1]
        weights = torch.softmax(weights.squeeze(-1), dim=0)  # [N]
        pooled = (image_embeddings * weights.unsqueeze(-1)).sum(dim=0)  # [D]
        return pooled

class PatentImageEncoder(nn.Module):
    def __init__(self, pooling='mean'):
        super().__init__()

        # Load pretrained ViT
        self.vit = ViTModel.from_pretrained('google/vit-base-patch16-224-in21k')
        self.hidden_size = self.vit.config.hidden_size  # e.g. 768

        # Choose pooling strategy
        if pooling == 'mean':
            self.pooling = self.mean_pooling
        elif pooling == 'weighted':
            self.pooling = WeightedImagePooling(self.hidden_size)
        else:
            raise ValueError("pooling must be 'mean' or 'weighted'")

    def mean_pooling(self, image_embeddings):  # [N, D]
        return image_embeddings.mean(dim=0)

    def forward(self, image_batch):  # image_batch: [N, 3, 224, 224] (N images for one patent)
        outputs = self.vit(pixel_values=image_batch)
        cls_embeddings = outputs.last_hidden_state[:, 0, :]  # [N, D]
        pooled = self.pooling(cls_embeddings)  # [D]
        return pooled


encoder = PatentImageEncoder()
encoder.eval()
features = []

with torch.no_grad():
    for batch_images, batch_ids in tqdm(loader):
        batch_features = []
        for image_tensor in batch_images:
            feature = encoder(image_tensor)  # [768]
            batch_features.append(feature)

        batch_features = torch.stack(batch_features)  # [batch_size, 768]
        print("Batch patent feature shape:", batch_features.shape)
        break
```

```python
def collate_fn(batch):
    # Extract all components from the batch
    titles = [item["title"] for item in batch]  # List of tokenized title chunks
    abstracts = [item["abstract"] for item in batch]  # List of tokenized abstract chunks
    claims = [item["claims"] for item in batch]  # List of tokenized claims chunks
    images = [item["images"] for item in batch]  # List of image tensors
    labels = [item["label"] for item in batch]  # List of label tensors

    # Handle text fields (title, abstract, claims)
    def process_text_field(field_data):
        # field_data is a list where each item is a dict with "input_ids" and "attention_mask"
        # Flatten all chunks across all samples
        all_input_ids = []
        all_attention_masks = []
        chunk_counts = []  # Track number of chunks per sample
        
        for sample in field_data:
            chunk_counts.append(len(sample["input_ids"]))
            all_input_ids.extend(sample["input_ids"])
            all_attention_masks.extend(sample["attention_mask"])
        
        # Stack all chunks
        input_ids = torch.stack(all_input_ids)
        attention_mask = torch.stack(all_attention_masks)
        
        return {
            "input_ids": input_ids,
            "attention_mask": attention_mask,
            "chunk_counts": torch.tensor(chunk_counts, dtype=torch.long)
        }

    processed_titles = process_text_field(titles)
    processed_abstracts = process_text_field(abstracts)
    processed_claims = process_text_field(claims)

    # Handle images to match text processing style
    def process_images(image_data):
        all_images = []
        image_counts = []  # Track number of images per sample
        
        for sample in image_data:
            num_images = sample.size(0)
            image_counts.append(num_images)
            if num_images > 0:
                all_images.append(sample)
        
        # Stack all images if there are any, otherwise return empty tensor
        if len(all_images) > 0:
            images_tensor = torch.cat(all_images, dim=0)
        else:
            # Create empty tensor with correct dimensions
            images_tensor = torch.empty(0, *image_data[0].shape[1:])
        
        return {
            "images": images_tensor,
            "image_counts": torch.tensor(image_counts, dtype=torch.long)
        }

    processed_images = process_images(images)

    return {
        "title": processed_titles,
        "abstract": processed_abstracts,
        "claims": processed_claims,
        "images": processed_images,
        "label": torch.stack(labels)
    }
```

```python
import torch
import torch.nn as nn
from transformers import AutoModel
from tqdm import tqdm

def set_encoder_trainable(model, trainable):
    """Sets all encoders to be trainable or frozen."""
    for encoder in [model.encoder_title, model.encoder_abstract, model.encoder_claims]:
        for param in encoder.parameters():
            param.requires_grad = trainable
class AttentionPooling(nn.Module):
    def __init__(self, hidden_size):
        super().__init__()
        self.attention = nn.Sequential(
            nn.Linear(hidden_size, hidden_size),
            nn.Tanh(),
            nn.Linear(hidden_size, 1)
        )
    def forward(self, token_embeddings, attention_mask):
        # attention_mask: [batch_size, seq_len]
        scores = self.attention(token_embeddings).squeeze(-1)  # [batch_size, seq_len]
        scores = scores.masked_fill(attention_mask == 0, -1e9)  # mask padded tokens
        weights = torch.softmax(scores, dim=1).unsqueeze(-1)    # [batch_size, seq_len, 1]
        return torch.sum(token_embeddings * weights, dim=1)     # weighted sum over tokens

class PatentEmbeddingModel(nn.Module):
    def __init__(self, device, hidden=128, model_name="bert-base-uncased", num_outputs=1, freeze_encoder=True):
        super().__init__()
        self.device = device
        self.hidden = hidden
        
        # Separate encoders for each input
        self.encoder_title = AutoModel.from_pretrained(model_name)
        self.encoder_abstract = AutoModel.from_pretrained(model_name)
        self.encoder_claims = AutoModel.from_pretrained(model_name)
        # self.encoder_description = AutoModel.from_pretrained(model_name)

        self.hidden_size = self.encoder_title.config.hidden_size

        # Projection layers for each text component
        self.proj_title = nn.Linear(self.hidden_size, self.hidden)
        self.proj_abstract = nn.Linear(self.hidden_size, self.hidden)
        self.proj_claims = nn.Linear(self.hidden_size, self.hidden)
        self.proj_description = nn.Linear(self.hidden_size, self.hidden)
        self.pooling = AttentionPooling(self.hidden_size)
        # Final regression layer
        self.regressor = nn.Linear(self.hidden * 3, num_outputs)
        # self.regressor = nn.Linear(self.hidden * 4, num_outputs)


        # Set encoder freeze state
        set_encoder_trainable(self, not freeze_encoder)

    def mean_pooling(self, model_output, attention_mask):
        token_embeddings = model_output.last_hidden_state
        mask_expanded = attention_mask.unsqueeze(-1).expand(token_embeddings.shape)
        sum_embeddings = torch.sum(token_embeddings * mask_expanded, dim=1)
        sum_mask = torch.clamp(mask_expanded.sum(dim=1), min=1e-9)
        return sum_embeddings / sum_mask

    def get_embedding(self, encoder, batched_inputs):
            
        input_ids = batched_inputs["input_ids"].to(self.device)
        attention_mask = batched_inputs["attention_mask"].to(self.device)
        chunk_counts = batched_inputs["chunk_counts"].to(self.device)
        
        # Get embeddings for all chunks
        with torch.no_grad():
            outputs = encoder(input_ids=input_ids, attention_mask=attention_mask)
        
        # Pool embeddings for each chunk (using your specified pooling method)
        chunk_embeddings = self.pooling(outputs.last_hidden_state, attention_mask)
        
        # Split chunk embeddings back to original samples using chunk_counts
        embeddings_split = torch.split(chunk_embeddings, chunk_counts.tolist())
        sample_embeddings = torch.stack([emb.mean(dim=0) for emb in embeddings_split])
        return sample_embeddings

    def forward(self, title, abstract, claims):
        # description_emb = self.proj_description(self.get_embedding(self.encoder_description, description))
        title_emb = self.proj_title(self.get_embedding(self.encoder_title, title))
        abstract_emb = self.proj_abstract(self.get_embedding(self.encoder_abstract, abstract))
        claims_emb = self.proj_claims(self.get_embedding(self.encoder_claims, claims))
        # description_emb = self.proj_description(self.get_embedding(self.encoder_description, description))
        # combined_emb = torch.cat([title_emb, abstract_emb, claims_emb, description_emb], dim=1)
        combined_emb = torch.cat([title_emb, abstract_emb, claims_emb], dim=1)
        return self.regressor(combined_emb)

```

```python
import torch
import torch.nn as nn
from tqdm import tqdm
import os

def train_model(model, device, train_loader, val_loader, epochs=3, lr=1e-4, fine_tune=False, save_path="best_model.pt"):
    # Adjust encoder trainability before training
    set_encoder_trainable(model, fine_tune)

    if fine_tune:
        optimizer = torch.optim.AdamW(model.parameters(), lr=lr)
    else:
        encoder_params = list(model.encoder_title.parameters()) + \
                         list(model.encoder_abstract.parameters()) + \
                         list(model.encoder_claims.parameters())
        encoder_param_ids = set(id(p) for p in encoder_params)
        non_encoder_params = [p for p in model.parameters() if id(p) not in encoder_param_ids]
        optimizer = torch.optim.AdamW(non_encoder_params, lr=lr)

    criterion = nn.MSELoss()
    model.train()
    best_val_loss = float("inf")

    for epoch in range(epochs):
        total_loss = 0
        model.train()
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs} [Training]", leave=False)

        for batch in progress_bar:
            optimizer.zero_grad()
            outputs = model(
                title=batch["title"],
                abstract=batch["abstract"],
                claims=batch["claims"]
            ).squeeze()
            loss = criterion(outputs, batch["label"].to(device))
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
            progress_bar.set_postfix(train_loss=loss.item())

        train_loss = total_loss / len(train_loader)

        # Evaluate on validation set
        model.eval()
        val_loss = 0
        with torch.no_grad():
            for batch in val_loader:
                outputs = model(
                    title=batch["title"],
                    abstract=batch["abstract"],
                    claims=batch["claims"]
                ).squeeze()
                loss = criterion(outputs, batch["label"].to(device))
                val_loss += loss.item()

        val_loss /= len(val_loader)
        print(f"Epoch {epoch+1}, Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}")

        # Save model if validation improves
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            torch.save(model.state_dict(), save_path)
            print(f"✅ New best model saved at epoch {epoch+1} with val loss {val_loss:.4f}")
```
